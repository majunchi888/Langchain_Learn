```python
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage, SystemMessage,AIMessage
from langchain_core.prompts import ChatPromptTemplate

from langchain_core.output_parsers import StrOutputParser


# Load environment variables from .env file
load_dotenv()

llm_openai = ChatOpenAI(model = "deepseek-chat",
base_url="https://api.deepseek.com",
api_key=os.environ.get('DEEPSEEK_API_KEY'),
 temperature=0)


if os.environ.get('DEEPSEEK_API_KEY') :
    print("DEEPSEEK_API_KEY is set")
else:    print("DEEPSEEK_API_KEY is not set")


```

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver 

agent = create_agent(
    model=llm_openai,
    checkpointer=InMemorySaver(),
    )
```

```python
config = {"configurable": {"thread_id": "thread_1"}}

response = agent.invoke(
    {"messages": [HumanMessage(content="你好，我叫马俊驰，我喜欢猫猫")]},
    config
 )
print(response)
```

```python
response = agent.invoke(
{"messages": [HumanMessage(content="我最喜欢的动物是什么")]},
config
)
print(response)
```

```python
import sqlite3
from langgraph.checkpoint.sqlite import SqliteSaver

#连接sqlite
connection = sqlite3.connect("SalesDB/checkpoint.db",check_same_thread=False)
#初始化checkpointer
checkpointer = SqliteSaver(connection)
#自动建表
checkpointer.setup()

agent2 = create_agent(model=llm_openai, checkpointer=checkpointer)
```

```python
config2 = {"configurable": {"thread_id": "thread_2"}}

response = agent2.invoke(
    {"messages": [HumanMessage(content="你好，我叫马俊驰，我喜欢猫猫")]},
    config2
 )
print(response)
```

```python
response = agent2.invoke(
{"messages": [HumanMessage(content="我最喜欢的动物是什么")]},
config2
)
print(response)
```

```python
from langchain.agents.middleware import SummarizationMiddleware

checkpointer = InMemorySaver()

middleware = SummarizationMiddleware(
    model=llm_openai,
    trigger = ("messages",3),
    keep=("messages",1),
    )

agent=create_agent(
    model=llm_openai,
    checkpointer=checkpointer,
    middleware=[middleware]
)

config = {"configurable": {"thread_id": "thread_3"}}

agent.invoke(
    {"messages": [HumanMessage(content="你好，我叫马俊驰")]},
    config
 )
agent.invoke(
    {"messages": [HumanMessage(content="我喜欢篮球")]},
    config
 )
agent.invoke(
    {"messages": [HumanMessage(content="我喜欢猫猫")]},
    config
 )
final_response=agent.invoke(
    {"messages": [HumanMessage(content="你还记得我吗")]},
    config
 )
```

```python
final_answer = final_response["messages"][-1].content 
print(final_answer)
```
